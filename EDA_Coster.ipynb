# Step 2: Data Preparation

In this step we will:
- Select relevant columns
- Convert dates
- Rename columns
- Check missing values
- Check duplicate records
- Remove duplicate roller coasters

```python
df = df[
    [
        'coaster_name',
        'Location',
        'Status',
        'Manufacturer',
        'year_introduced',
        'latitude',
        'longitude',
        'Type_Main',
        'opening_date_clean',
        'speed_mph',
        'height_ft',
        'Inversions_clean',
        'Gforce_clean'
    ]
].copy()
```

```python
df.head()
```

```python
df['opening_date_clean'] = pd.to_datetime(
    df['opening_date_clean'],
    errors='coerce'
)
```

```python
df.dtypes
```

```python
df = df.rename(
    columns={
        'coaster_name': 'Coaster_Name',
        'year_introduced': 'Year_Introduced',
        'opening_date_clean': 'Opening_Date',
        'speed_mph': 'Speed_mph',
        'height_ft': 'Height_ft',
        'Inversions_clean': 'Inversions',
        'Gforce_clean': 'Gforce'
    }
)
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False
)

plt.title('Missing Values in Dataset')
plt.show()
```

```python
df.loc[df.duplicated()]
```

```python
df.loc[
    df.duplicated(subset=['Coaster_Name'])
].head(10)
```

```python
df.query(
    'Coaster_Name == "Crystal Beach Cyclone"'
)
```

```python
df = df.loc[
    ~df.duplicated(
        subset=[
            'Coaster_Name',
            'Location',
            'Opening_Date'
        ]
    )
].reset_index(drop=True).copy()
```

# Step 3: Feature Understanding

Univariate analysis examines individual variables.

We will use:
- Value counts
- Bar charts
- Histograms
- KDE plots
- Boxplots

```python
df['Year_Introduced'].value_counts().head(10)
```

```python
ax = (
    df['Year_Introduced']
    .value_counts()
    .head(10)
    .plot(
        kind='bar',
        figsize=(10, 5),
        title='Top 10 Years Coasters Were Introduced'
    )
)

ax.set_xlabel('Year Introduced')
ax.set_ylabel('Number of Coasters')

plt.xticks(rotation=45)
plt.show()
```

```python
df['Speed_mph'].describe()
```

```python
ax = df['Speed_mph'].plot(
    kind='hist',
    bins=20,
    figsize=(10, 5),
    title='Coaster Speed Distribution'
)

ax.set_xlabel('Speed (mph)')
ax.set_ylabel('Number of Coasters')

plt.show()
```

```python
ax = df['Speed_mph'].plot(
    kind='kde',
    figsize=(10, 5),
    title='Coaster Speed Distribution'
)

ax.set_xlabel('Speed (mph)')

plt.show()
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(x=df['Speed_mph'])

plt.title('Boxplot of Coaster Speed')
plt.xlabel('Speed (mph)')

plt.show()
```

```python
df['Type_Main'].value_counts()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    y='Type_Main',
    order=df['Type_Main'].value_counts().index
)

plt.title('Number of Coasters by Type')
plt.xlabel('Number of Coasters')
plt.ylabel('Coaster Type')

plt.show()
```

# Step 4: Feature Relationships

We will investigate relationships between:
- Speed and Height
- Year Introduced and Speed
- Height and Speed
- Inversions and Speed
- G-force and Speed

We will use:
- Scatter plots
- Pair plots
- Correlation matrix
- Heatmap

```python
df.plot(
    kind='scatter',
    x='Speed_mph',
    y='Height_ft',
    figsize=(10, 6),
    title='Coaster Speed vs Height'
)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

ax = sns.scatterplot(
    data=df,
    x='Speed_mph',
    y='Height_ft',
    hue='Year_Introduced'
)

ax.set_title('Coaster Speed vs Height')
ax.set_xlabel('Speed (mph)')
ax.set_ylabel('Height (ft)')

plt.show()
```

```python
sns.pairplot(
    df,
    vars=[
        'Year_Introduced',
        'Speed_mph',
        'Height_ft',
        'Inversions',
        'Gforce'
    ],
    hue='Type_Main'
)

plt.show()
```

```python
df_corr = df[
    [
        'Year_Introduced',
        'Speed_mph',
        'Height_ft',
        'Inversions',
        'Gforce'
    ]
].dropna().corr()

df_corr
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    df_corr,
    annot=True,
    fmt='.2f'
)

plt.title('Correlation Matrix')
plt.show()
```

```python
location_speed = (
    df.query('Location != "Other"')
    .groupby('Location')['Speed_mph']
    .agg(['mean', 'count'])
    .query('count >= 10')
    .sort_values('mean')
)

location_speed
```

```python
ax = (
    location_speed['mean']
    .plot(
        kind='barh',
        figsize=(12, 8),
        title='Average Coaster Speed by Location'
    )
)

ax.set_xlabel('Average Coaster Speed (mph)')
ax.set_ylabel('Location')

plt.show()
```

```python
location_speed.sort_values(
    'mean',
    ascending=False
).head(10)
```

```python
df[
    [
        'Coaster_Name',
        'Location',
        'Speed_mph',
        'Height_ft',
        'Year_Introduced'
    ]
].sort_values(
    'Speed_mph',
    ascending=False
).head(10)
```

```python
df[
    [
        'Coaster_Name',
        'Location',
        'Height_ft',
        'Speed_mph',
        'Year_Introduced'
    ]
].sort_values(
    'Height_ft',
    ascending=False
).head(10)
```

```python
df[
    [
        'Coaster_Name',
        'Location',
        'Inversions',
        'Speed_mph',
        'Height_ft'
    ]
].sort_values(
    'Inversions',
    ascending=False
).head(10)
```

```python
df.groupby('Type_Main')[
    [
        'Speed_mph',
        'Height_ft',
        'Inversions',
        'Gforce'
    ]
].mean().sort_values(
    'Speed_mph',
    ascending=False
)
```

7. Location-level analysis can be used to compare average coaster speeds,

   more meaningful.

8. The analysis demonstrates how statistical summaries and visualizations
